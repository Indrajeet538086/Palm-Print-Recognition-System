# Normalisation

```python
import cv2
import os
import numpy as np
from pathlib import Path
```

```python
TUNING_CONFIG = {
    "input_blur_dir": Path(r"C:\Users\singh\Users\singh\Desktop\Palm detection\ROI\04_Noise_Reduced 7 latest work_7-3"),
    "output_norm_dir": Path(r"C:\Users\singh\Users\singh\Desktop\Palm detection\ROI\Normalized image")
}

# Ensure final destination folder exists
os.makedirs(TUNING_CONFIG["output_norm_dir"], exist_ok=True)

if __name__ == "__main__":
    # Scan the Blur folder for all images recursively
    blurred_images = list(TUNING_CONFIG["input_blur_dir"].rglob('*.png'))
    
    if not blurred_images:
        print(f" No images found in {TUNING_CONFIG['input_blur_dir']}. Ensure Step 4 ran successfully!")
    else:
        print(f" Starting separate Min-Max Normalization for {len(blurred_images)} items...")
        processed_count = 0
        
        for blur_img_path in blurred_images:
            # Read image directly as a 1-channel grayscale matrix
            blurred_roi = cv2.imread(str(blur_img_path), cv2.IMREAD_GRAYSCALE)
            if blurred_roi is None: 
                continue
            
            # PURE NORMALIZATION: Stretch pixel intensities uniformly across the full 0-255 spectrum
            # This standardizes contrast dynamically across different lighting environments
            normalized_roi = cv2.normalize(
                src=blurred_roi, 
                dst=None, 
                alpha=0, 
                beta=255, 
                norm_type=cv2.NORM_MINMAX, 
                dtype=cv2.CV_8U
            )
            
            # Reconstruct exact identical sub-folder trees (User_ID / Handedness)
            relative_structure = blur_img_path.relative_to(TUNING_CONFIG["input_blur_dir"])
            save_path = TUNING_CONFIG["output_norm_dir"] / relative_structure
            
            # Ensure sub-folders exist
            os.makedirs(save_path.parent, exist_ok=True)
            
            # Save the normalized file
            cv2.imwrite(str(save_path), normalized_roi)
            processed_count += 1
            
        print("=" * 55)
        print(f" Stage 5 Complete. Image normalization finished.")
        print(f"Total normalized outputs saved: {processed_count}")
        print(f"Final Save Path: {TUNING_CONFIG['output_norm_dir']}")
        print("=" * 55)
```
